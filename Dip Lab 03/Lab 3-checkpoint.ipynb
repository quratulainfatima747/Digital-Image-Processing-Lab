## Task 1: Brightness Adjustment

**Objective:** To increase and decrease the brightness of a grayscale image.

**Formula:**

s = r + β

Where:

* r = Original pixel intensity
* β = Brightness value
* s = New pixel intensity

We use:

* β = +40 for brighter image.
* β = -40 for darker image.

Pixel values must remain between 0 and 255.


### Dry Run

Given pixels:

[30, 80, 128, 180, 240]

**1. Brightness +40**

Formula: s = r + 40

Raw Values:

[70, 120, 168, 220, 280]

After Clipping:

[70, 120, 168, 220, 255]

**2. Brightness -40**

Formula: s = r - 40

Raw Values:

[-10, 40, 88, 140, 200]

After Clipping:

[0, 40, 88, 140, 200]

**Observation:**

Pixel values greater than 255 become 255, while values below 0 become 0. This process is called clipping. It can cause loss of image details.


```python
# Brightness Adjustment using NumPy

# Convert image into int16
img = image.astype(np.int16)

# Increase brightness by 40
bright = np.clip(img + 40, 0, 255).astype(np.uint8)

# Decrease brightness by 40
dark = np.clip(img - 40, 0, 255).astype(np.uint8)

print("Brightness adjustment completed!")
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(image, cmap='gray', vmin=0, vmax=255)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(bright, cmap='gray', vmin=0, vmax=255)
plt.title("Brighter Image (+40)")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(dark, cmap='gray', vmin=0, vmax=255)
plt.title("Darker Image (-40)")
plt.axis('off')

plt.show()
```

```python
# Brightness adjustment using nested loops

loop_bright = np.zeros(image.shape, dtype=np.uint8)

for i in range(image.shape[0]):
    for j in range(image.shape[1]):

        value = int(image[i, j]) + 40

        if value > 255:
            value = 255

        loop_bright[i, j] = value

print("Nested loops completed!")

```

```python
# Compare NumPy and Nested Loops

if np.array_equal(bright, loop_bright):
    print("Both results are identical!")
else:
    print("Results are different!")
```

### Observation

Brightness adjustment changes the intensity of image pixels. Adding 40 makes the image brighter, while subtracting 40 makes it darker. Pixel values are clipped between 0 and 255 to prevent invalid intensity values. The NumPy and nested loop implementations produce identical results.


## Task 2: Linear Contrast from Zero

**Objective:** To increase image contrast using a linear intensity transformation.

**Formula:**

s = αr + β

Where:

* s = Output pixel intensity
* r = Original pixel intensity
* α = Contrast factor (1.5)
* β = Brightness offset (0)

In this task, we use α = 1.5 and β = 0 to increase image contrast.


### Dry Run

Given pixels:

[30, 80, 128, 180, 240]

Formula:

s = 1.5 × r

Calculations:

30 × 1.5 = 45

80 × 1.5 = 120

128 × 1.5 = 192

180 × 1.5 = 270

240 × 1.5 = 360

**Raw Values:**

[45, 120, 192, 270, 360]

**After Clipping (0–255):**

[45, 120, 192, 255, 255]

**Observation:**

The contrast factor 1.5 increases pixel intensity values. Values greater than 255 are clipped to 255. Pixels 180 and 240 become saturated, causing loss of highlight details.


```python
# Linear Contrast from Zero

alpha = 1.5
beta = 0

# Convert image into float32
img = image.astype(np.float32)

# Apply linear contrast
contrast = alpha * img + beta

# Clip values between 0 and 255
contrast = np.clip(contrast, 0, 255)

# Convert into uint8
contrast = contrast.astype(np.uint8)

print("Linear contrast applied successfully!")
```

```python
plt.figure(figsize=(12, 5))

# Original Image
plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray', vmin=0, vmax=255)
plt.title("Original Image")
plt.axis('off')

# Contrast Enhanced Image
plt.subplot(1, 2, 2)
plt.imshow(contrast, cmap='gray', vmin=0, vmax=255)
plt.title("Linear Contrast (Alpha=1.5)")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
pixels = np.array([30, 80, 128, 180, 240])

raw_values = 1.5 * pixels

final_values = np.clip(raw_values, 0, 255).astype(np.uint8)

print("Original Pixels:", pixels)

print("Raw Values:", raw_values)

print("Final Values:", final_values)
```

### Observation

Linear contrast increases the intensity values of the image using a contrast factor of 1.5. The enhanced image appears brighter and shows increased intensity levels. Values exceeding 255 are clipped to 255, resulting in loss of some highlight details.

## Task 5: Gamma Correction

Gamma correction is used to adjust the brightness of an image.

Formula:

s = 255 × (r / 255)^γ

We will apply two gamma values:
- Gamma = 0.5
- Gamma = 2.0

Gamma less than 1 makes the image brighter, while gamma greater than 1 makes the image darker.

### Dry Run

Input pixels: [0, 64, 128, 192, 255]

Formula:
s = 255 × (r / 255)^γ

For γ = 0.5:

Output = [0, 127.749, 180.665, 221.269, 255]

After truncation:
[0, 127, 180, 221, 255]

For γ = 2.0:

Output = [0, 16.062, 64.251, 144.564, 255]

## Task 7: Global Histogram Equalization

Histogram equalization is an image enhancement technique used to improve image contrast.

It redistributes the intensity values across the available range (0–255).

OpenCV provides the `equalizeHist()` function to perform global histogram equalization.

Formula:

s = (L - 1) × CDF(r)

Where:
- L = Number of intensity levels (256)
- CDF = Cumulative Distribution Function
- r = Original intensity value

### Dry Run

Consider the following pixel values:

Input pixels: [50, 50, 100, 100, 150, 150, 200, 200]

Frequency of each intensity:

50  = 2
100 = 2
150 = 2
200 = 2

Cumulative frequencies:

50  = 2
100 = 4
150 = 6
200 = 8

Total pixels = 8

CDF values:

50  = 2/8 = 0.25
100 = 4/8 = 0.50
150 = 6/8 = 0.75
200 = 8/8 = 1.00

Using the mapping formula:

s = 255 × CDF

Mapped values (before rounding):

50  → 63.75
100 → 127.50
150 → 191.25
200 → 255

After truncation:

50  → 63
100 → 127
150 → 191
200 → 255

```python
pixels = np.array([50, 50, 100, 100, 150, 150, 200, 200])

unique_values, counts = np.unique(pixels, return_counts=True)
cdf = np.cumsum(counts) / len(pixels)

print("Unique Intensities:", unique_values)
print("Frequencies:", counts)
print("CDF:", cdf)
print("Mapped Values:", np.trunc(255 * cdf).astype(np.uint8))
```

```python
equalized_image = cv2.equalizeHist(image)

print("Histogram Equalization Completed!")
print("Original Shape:", image.shape)
print("Equalized Shape:", equalized_image.shape)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray', vmin=0, vmax=255)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(equalized_image, cmap='gray', vmin=0, vmax=255)
plt.title("Global Histogram Equalization")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
original_hist = cv2.calcHist([image], [0], None, [256], [0, 256])
equalized_hist = cv2.calcHist([equalized_image], [0], None, [256], [0, 256])

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(original_hist, color='black')
plt.title("Original Histogram")
plt.xlabel("Intensity")
plt.ylabel("Pixel Count")
plt.xlim([0, 256])

plt.subplot(1, 2, 2)
plt.plot(equalized_hist, color='black')
plt.title("Equalized Histogram")
plt.xlabel("Intensity")
plt.ylabel("Pixel Count")
plt.xlim([0, 256])

plt.tight_layout()
plt.show()
```

### Observation

Global histogram equalization redistributes image intensity values to improve contrast.

The equalized image may show more visible details than the original image, especially when the original image has low contrast.

The histogram also shows how pixel intensities are redistributed after equalization.

## Task 8: CLAHE (Contrast Limited Adaptive Histogram Equalization)

CLAHE is an image enhancement technique used to improve local contrast.

It divides an image into small regions called tiles and applies histogram equalization to each tile.

The clip limit prevents excessive contrast enhancement and reduces noise amplification.

Parameters:
- clipLimit = 2.0
- tileGridSize = (8, 8)

### Dry Run

Consider an image divided into 4 tiles:

Tile 1: Dark region
Tile 2: Bright region
Tile 3: Low contrast region
Tile 4: High contrast region

CLAHE processes each tile separately.

1. Divide the image into tiles.
2. Calculate the histogram of each tile.
3. Limit excessive histogram peaks using the clip limit.
4. Redistribute the clipped histogram values.
5. Improve local contrast.
6. Combine the processed tiles into one image.

Parameters:
clipLimit = 2.0
tileGridSize = (8, 8)

Expected result:
Local contrast and visibility of image details improve.

```python
clahe = cv2.createCLAHE(
    clipLimit=2.0,
    tileGridSize=(8, 8)
)

clahe_image = clahe.apply(image)

print("CLAHE Applied Successfully!")
print("Original Image Shape:", image.shape)
print("CLAHE Image Shape:", clahe_image.shape)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray', vmin=0, vmax=255)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(clahe_image, cmap='gray', vmin=0, vmax=255)
plt.title("CLAHE Enhanced Image")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
original_hist = cv2.calcHist([image], [0], None, [256], [0, 256])
clahe_hist = cv2.calcHist([clahe_image], [0], None, [256], [0, 256])

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(original_hist, color='black')
plt.title("Original Histogram")
plt.xlabel("Intensity")
plt.ylabel("Pixel Count")
plt.xlim([0, 256])

plt.subplot(1, 2, 2)
plt.plot(clahe_hist, color='black')
plt.title("CLAHE Histogram")
plt.xlabel("Intensity")
plt.ylabel("Pixel Count")
plt.xlim([0, 256])

plt.tight_layout()
plt.show()
```

### Observation

CLAHE improves local contrast by processing different regions of the image separately.

The clip limit controls excessive contrast enhancement and helps reduce noise amplification.

The resulting image may show more visible details in dark and low-contrast regions.

## Task 9: Fixed Thresholding

Fixed thresholding is an image segmentation technique.

It converts a grayscale image into a binary image using a fixed threshold value.

Formula:

If r > T, then s = 255
If r <= T, then s = 0

Where:
- r = Original pixel intensity
- T = Threshold value
- 255 = White
- 0 = Black

We will use threshold value T = 127.

### Dry Run

Input pixels:

[30, 80, 128, 180, 240]

Threshold value:

T = 127

Apply the condition:

30  <= 127  → 0
80  <= 127  → 0
128 > 127   → 255
180 > 127   → 255
240 > 127   → 255

Final Output:

[0, 0, 255, 255, 255]

```python
pixels = np.array([30, 80, 128, 180, 240], dtype=np.uint8)

threshold_value = 127

result = np.where(pixels > threshold_value, 255, 0).astype(np.uint8)

print("Original Pixels:", pixels)
print("Threshold Value:", threshold_value)
print("Thresholded Pixels:", result)
```

```python
threshold_value = 127

ret, threshold_image = cv2.threshold(
    image,
    threshold_value,
    255,
    cv2.THRESH_BINARY
)

print("Fixed Thresholding Applied Successfully!")
print("Threshold Value:", ret)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray', vmin=0, vmax=255)
plt.title("Original Grayscale Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(threshold_image, cmap='gray', vmin=0, vmax=255)
plt.title("Fixed Thresholding (T = 127)")
plt.axis('off')

plt.tight_layout()
plt.show()
```

### Observation

Fixed thresholding converts a grayscale image into a binary image.

- Pixel values greater than 127 become white (255).
- Pixel values less than or equal to 127 become black (0).

This technique separates brighter regions from darker regions of an image.

## Task 11: Adaptive Gaussian Thresholding

Adaptive Gaussian Thresholding is an image segmentation technique.

It calculates a different threshold for each small region of an image instead of using one fixed threshold.

The threshold is calculated using the weighted sum of neighboring pixels with Gaussian weighting.

Parameters:
- maxValue = 255
- adaptiveMethod = ADAPTIVE_THRESH_GAUSSIAN_C
- thresholdType = THRESH_BINARY
- blockSize = 11
- C = 2

Block size must be an odd number greater than 1.

### Dry Run

Consider a small image region with the following pixel values:

[40, 50, 60]
[50, 80, 70]
[60, 70, 90]

Adaptive Gaussian Thresholding calculates a weighted local mean using Gaussian weights.

Formula:

T(x,y) = Weighted Local Mean - C

Where:
- T(x,y) = Local threshold
- C = Constant value
- Weighted Local Mean = Gaussian-weighted average of neighboring pixels

For example, if the weighted local mean is 65 and C = 2:

Threshold = 65 - 2 = 63

For a pixel value of 80:

80 > 63, so output = 255.

For a pixel value of 50:

50 <= 63, so output = 0.

The actual threshold is calculated separately for each pixel using its neighborhood.

```python
sample = np.array([
    [40, 50, 60],
    [50, 80, 70],
    [60, 70, 90]
], dtype=np.uint8)

sample_blurred = cv2.GaussianBlur(sample, (3, 3), 0)

sample_threshold = np.clip(
    sample_blurred.astype(np.int16) - 2,
    0, 255
).astype(np.uint8)

sample_result = np.where(
    sample > sample_threshold, 255, 0
).astype(np.uint8)

print("Original Sample:\n", sample)
print("Gaussian Weighted Values:\n", sample_blurred)
print("Local Thresholds:\n", sample_threshold)
print("Binary Result:\n", sample_result)
```

```python
adaptive_gaussian = cv2.adaptiveThreshold(
    image,
    255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY,
    11,
    2
)

print("Adaptive Gaussian Thresholding Applied Successfully!")
print("Original Image Shape:", image.shape)
print("Result Image Shape:", adaptive_gaussian.shape)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray', vmin=0, vmax=255)
plt.title("Original Grayscale Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(adaptive_gaussian, cmap='gray', vmin=0, vmax=255)
plt.title("Adaptive Gaussian Thresholding")
plt.axis('off')

plt.tight_layout()
plt.show()
```

### Observation

Adaptive Gaussian Thresholding converts a grayscale image into a binary image.

It calculates local threshold values for different image regions using Gaussian weighting.

This method is useful for images with uneven lighting because each region gets its own threshold.

The output contains black (0) and white (255) pixels.
